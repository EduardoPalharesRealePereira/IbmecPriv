Newton Raphson 
- É um método de encontrar a raiz de uma equação de forma aberta, ja que ele não "encurrala" a raiz dando inicio e fim, como na bissecção e regula falsi, ele da apenas um chute (xi) e faz uma reta tangente com esse ponto, ai onde essa reta tangente cruzar com o eixo x, ele faz uma nova estimativa a partir dali.
- Costuma convergir mais rapido que os metodos fechados, porém, como nem tudo são flores, corre mais risco de divergir, (xi+1 vai pra longe da raiz, se afastando cada vez mais), não tendo resultado correto.
- newton(funcao,chute_inicial,derivada_da_funcao)
- newton(f,x0,df)



```python
from numpy import linspace
from scipy.optimize import newton
from matplotlib.pyplot import plot, legend, grid, xlabel, ylabel

def f(x):
    y = x**3 - 6*x**2 + 11*x - 6.1
    return y


def df(x):
    z = 3*x**2 - 12*x + 11
    return z

x = linspace(0,4,100) #se eu diminuir o tamanho do x da função aqui (que pode ser escolhido), pode ser que o ponto fique fora da função, caso ela seja menor que 3.05(que é o valor da raiz(ponto))
x0 = 3.5

raiz = newton(f, x0, df)
print("raiz:",raiz)

plot(x, f(x), 'r', label="função")
plot(raiz, f(raiz), 'ko', label="raiz")
legend()
grid()

```
