Derivacao numerica: função gradient (centrada, regressiva e progressiva)

- gradient(f(x), x, edge_order = 2)
- Serve para: Estimar derivada quando você só tem pontos (tabela) ou não quer derivar na mão.
- Diferença pra Taylor: EU so vou precisar derivar se me pedir pra comparar o resultado com o valor verdadeiro (que é o da derivada)
- Centrada é a mais precisa. Quanto menor o h, menor o erro.
- voce cria malha de pontos com array
- Pontas do gradient são menos precisas que o miolo.

```python
from numpy import *
from matplotlib.pyplot import *


f = lambda x: 25*x**3 - 6*x**2 + 7*x - 88
df = lambda x: 75*x**2 - 12*x + 7

x = 2 
h = 0.2

xanterior = f(x-h)
xposterior = f(x+h)

regressiva = (f(x) - xanterior)/h
progressiva = (xposterior - f(x))/h
centrada = (xposterior - xanterior)/(2*h)
centrada2deriv = (xposterior - 2*f(x) + xanterior) / h**2


reg = round(regressiva, 4)
prog = round(progressiva,4)
cen = round(centrada, 4)
cen2 = round(centrada2deriv, 4)


ErroReg = ((df(x) - regressiva)/df(x)) * 100
ErroProg = ((df(x) - progressiva)/df(x)) * 100
ErroCen = ((df(x) - centrada)/df(x)) * 100
ErroCen2 = ((df(x) - centrada2deriv)/df(x)) * 100


print(f"O resultado exato é: ", df(x))
print(f"O resultado da regressiva é: ", reg)
print(f"O resultado da progressiva é: ", prog)
print(f"O resultado da centrada é: ", cen)
print(f"O resultado da centrada 2ª derivada é: ", cen2)
print("===================")
print(f"Erro Regressiva: {round(ErroReg, 2)} %")
print(f"Erro Progressiva: {round(ErroProg, 2)} %")
print(f"Erro Centrada: {round(ErroCen, 2)} %")
print(f"Erro Centrada 2 derivada: {round(ErroCen2, 2)} %")

xx = linspace(1, 3, 100)
plot(xx, df(xx), 'g-', label="função")
plot(x, progressiva, 'bd', label="progressiva")
plot(x, regressiva, 'gd', label="regressiva")
plot(x, centrada, 'yd', label="centrada")
plot(x, df(x), 'ks', label="exata")
legend()
grid()
```
