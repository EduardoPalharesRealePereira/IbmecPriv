Newton(interpolação) - (polyfit) -> n pontos = polinomio de grau n-1
exemplos:
reta -> 2 pontos (grau 1)
parabola -> 3 pontos (grau 2)
polinômio -> 4 pontos (grau 3)

polyfit(x,b,grau) onde x e b são os 2 arrays (cada um com sua linha de pontos)

- eh mais eficiente que lagrange (principalmente caso precise adicionar um ponto novo), porem eh menos intuitiva

INTERPOLAR EH ACHAR UMA FUNCAO (SEJA DE RETA, PARABOLA OU POLINOMIO) COM BASE NOS DADOS QUE NOS SÃO DADOS (ARRAYS DE DADOS), E COM BASE NESSA FUNÇÃO ACHAR O VALOR DE Y NO PONTO X.

O Fenômeno de Runge (conceito teórico importante)
Esse tem nome próprio e ela dedica espaço, então é candidato a questão teórica. O problema:

Quando você usa um polinômio de grau alto (muitos pontos) pra interpolar num intervalo largo, o polinômio começa a oscilar violentamente, principalmente nas extremidades do intervalo. Ele passa pelos pontos, mas entre eles faz ondulações absurdas, se afastando da função real.

Ou seja: mais pontos nem sempre é melhor. Um polinômio de grau muito alto pode ficar péssimo entre os pontos. A função clássica que mostra isso é f(x) = 1/(1+25x²).

EXTRAPOLACAO: ALGO QUE OCORRE CASO TENTEMOS ESTIMAR O VALOR DE UM PONTO FORA DO INTERVALO CONHECIDO. QUANTO MAIS LONGE DO VALOR CONHECIDO, MAIOR A CHANCE DE ERRO (maior a chance de ir pra qualquer direção e virar algo muito errado)


```python
from numpy import array, polyfit, poly1d

x = array([8, 9, 11])
b = array([0.9030900, 0.9542425, 1.0413927])

A = polyfit(x, b, 2)   # acha os coeficientes do polinômio de grau 2
f = poly1d(A)          # monta o polinômio a partir dos coeficientes

print(f)
print("Valor estimado em x=10:", f(10))
```
