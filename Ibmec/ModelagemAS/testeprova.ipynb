Q2 - Misturando metodos (otimizacao de newton e minimize )

```python
from numpy import *
from scipy.optimize import *
from matplotlib.pyplot import *


def f(x):
    y = 2*sin(x)-x**2/10
    return -y

xi = 0
x2 = 4

resposta = minimize(f, 0)
resultado = resposta.x
ymaximo = -f(resultado)
print(resultado)

#agora por newton

def w(x):
    z = 2*sin(x)-x**2/10
    return z



def df(x):
    w = 2*cos(x) - x/5
    return w


respsecante = newton(df,1)
print(respsecante)
respnewton = newton(df,1,f)
print(respnewton)

# com isso podemos concluir que:
# com minimize, precisamos apenas da função (coloca o retorno negativo se for maximize), e do ponto inicial
# ai a partir disso fazemos: resposta.x para obter o x maximo e fazemos o caminho inverso para obter ymaximo (-f(resultado))
# e concluimos que para secante: precisamos da derivada da função e do ponto de y onde x é maximo 
# e para newton basta ter a função normal + o ponto y onde x eh maximo + a derivada da funcao

xx = linspace(0,4,100)
plot(xx, w(xx), 'r-', label="funcao")
plot(resultado,w(resultado), 'gs', label="maximize")
plot(respsecante,w(respsecante), 'yo', label="secante")
plot(respnewton,w(respnewton), 'k*', label="newton")

grid()
legend()
```
