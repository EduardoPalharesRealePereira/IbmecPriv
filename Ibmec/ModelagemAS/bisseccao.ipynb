Bissecção 
 - Você tem uma função e os pontos (xl da esquerda, xu da direita)
 - ponto medio deles (xl + xu)/2 é o valor de xr
 - Se o valor dos 2 (xl e xu) jogados na função derem valor invertido (1 negativo e 1 positivo), significa que tem raiz no intervalo
 - Pra ir ate o ponto de parada eu vou testando (xl * xr). Se for > 0, então agora xr = xl, se for < 0, agora xu = xl
 - Vou fazendo essas substituições ate chegar na condição de erro ou de iterações

```python
from numpy import *
from matplotlib.pyplot import *
from scipy.optimize import *


def f(x): 
    y = x**2 + x - 5
    return y

#chutando valores pra esquerda e pra direita da raiz
xl = 1
xu = 2

raiz = bisect(f,xl,xu, full_output=True)
xr=raiz[0]

print("Raiz encontrada em x: ",xr)
print("Valor da função: ", raiz)


```

```python
from numpy import *
from scipy.optimize import *
from matplotlib.pyplot import *

def f(x):
    y = x**3 - 10
    return y


xl = 2    
xu = 3

funcao = bisect(f,xl,xu, full_output=True)
xr = funcao[0]

print("Resultado da função: ", funcao)
print("=================")
print("raiz: ",xr)
```

Agora a mesma questão que essa ultima so que na mão, sem bissect

```python
from numpy import *
from matplotlib.pyplot import *
def f(x):
    y = x**3 - 10
    return y


xl = 2
xu = 3
erro = 0.01
it = 1
xr = (xl+xu)/2

while(abs(f(xr)) > erro):
    if(f(xl)*f(xr) > 0):
        xl = xr
    
    else:
        xu = xr
    xr = (xl+xu)/2
    it = it+1


print(f"Precisão atingida em {it} iterações")
print("Raiz encontrada em: ",xr)
print("O valor da função na raiz: ", f(xr))

xx = linspace(2,3,100)
plot(xx, f(xx), 'r-', label="funcao")
plot(xr,f(xr), 'gs', label="raiz")
legend()
grid()
```
