Secante

- Muito parecida com newton raphson (ela é aberta tambem), mas ao inves de utilizar apenas 1 chute + 1 derivada igual newton, ela utiliza 2 chutes (xi-1 e xi) e joga esses valores na formula
-Ao inves da secante utilizar um chute em uma derivada, ela faz uma aproximação do ponto na derivada. Utilizada quando temos problemas chatos de derivar na mao.
-Quando a secante vai andando pra frente, xi-1 vira o xi e xi vira xi+1
-Newton ainda converge mais rapido que a secante, ja que a convergência de newton é quadratica, ja a de secante é superlinear. Mas mesmo assim ela ainda é mais rapida que bisect e regula-falsi.

- No codigo, resumindo: utilizamos a função newton tambem, mas no parametro passamos apenas a função e o x1, sem a derivada, desse jeito:
 newton(f, x0)
Assim, o python ja entende implicitamente que é secante (por nao ter a derivada), e o segundo ponto é feito pela função automaticamente tb,(o segundo valor do chute inicial é feito multiplicando o x0 por 1,0001%, pra ter sentido)

```python
from numpy import linspace, log10, sqrt
from scipy.optimize import newton
from matplotlib.pyplot import plot, legend, grid, xlabel, ylabel

def karman(f):
    Re = 100000.0
    y = 4.0*log10(Re*sqrt(f)) - 0.4 - (1/sqrt(f))
    return y

x = linspace(0.002, 0.008, 100)
plot(x, karman(x), 'r', label='karman(f)')
xlabel('f')
ylabel('karman(f)')

x0 = 0.007
raiz = newton(karman, x0)
plot(raiz, karman(raiz), 'ko', label='raiz')
legend()
grid()
```
