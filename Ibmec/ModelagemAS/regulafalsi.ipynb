Regula falsi

- É basicamente igual a bissect, mas a formula muda, agora tem f(xl) e f(xu) na formula pra achar xr
- Na maioria das vezes é mais preciso e rapido que bissect, mas NEM SEMPRE. Tem situações em que um dos lados da regula falsi fica estagnado, a gente faz todas as iterações mudando so um lado, então ele acaba ficando mais lento que o bissect. Lembrando que na bissecção sempre vai cortando os 2 lados igual (tanto de xl quanto de xu).
