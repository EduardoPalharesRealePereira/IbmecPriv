gauss seidal tem um probleminha -> as vezes não converge ou quando converge pode ser muito (problema da diagonal dominante)
 -> Para saber se converge: DOMINANCIA DIAGONAL - Em cada linha, o valor absoluto do número da diagonal deve ser maior que a soma dos valores absolutos dos outros números da mesma linha.

```python
from scipy.sparse.linalg import cgs
from numpy import array

A = array([[10, 2, -1], [-3, -6, 2], [1, 1, 5]])
b = array([[27], [-61.5], [-21.5]])

resultado = cgs(A, b, [0, 0, 0])
print(resultado[0])
```
