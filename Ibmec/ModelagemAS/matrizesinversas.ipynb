matrizes inversas

Sistema vira A·x = b (coeficientes, incógnitas, resultados).
Declarar com array([[linha],[linha]]).
Multiplicação de matriz é dot(), nunca *.
Resolver: solve(A, b) ou dot(inv(A), b).
cond(A) mede confiabilidade (alto = mal-condicionada).
Resposta a estímulos: a inversa diz como cada saída reage a mudanças nas entradas.

incognitas = os cs da questao do reator (c1,c2,c3), a conta que o solve ou Inversa * b faz eh multiplicar o c de cada linha, exemplo: c1 = matrizdosolve[0][0]*b1 + matrizdosolve[1][0]*b2 + matrizdosolve[2][0]*b3
