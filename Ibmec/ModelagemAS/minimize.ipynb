Minimize

- Função pronta do scipy que acha o ponto de MÍNIMO de uma função automaticamente (a "irmã" do bisect e do newton, mas pra otimização em vez de raiz).
- Aparece na Aula 9 (otimização). Usa um chute inicial, igual ao newton: minimize(f, x0)
- O resultado vem num pacote; o ponto de mínimo fica no .x → minimo = resultado.x
- Só sabe MINIMIZAR. Pra achar um MÁXIMO, usa o truque Max f = Min -f (inverte o sinal da função e minimiza).
- Resumindo o código:
 resultado = minimize(f, x0)
 minimo = resultado.x

```python
from numpy import linspace
from scipy.optimize import minimize
from matplotlib.pyplot import plot, grid, legend, xlabel, ylabel

def D(V):
    sigma = 0.5
    W = 15000
    y = 0.01*sigma*V**2 + (0.95/sigma)*(W/V)**2
    return y

V = linspace(100, 1200, 100)
xinicial = 400

plot(V, D(V), 'b', label="Função de Arrasto")
xlabel('Velocidade')
ylabel('Arrasto')

resultado = minimize(D, xinicial)
minimo = resultado.x
print(minimo)

plot(minimo, D(minimo), "ro", label="Ponto de Mínimo")
legend()
grid()
print(f"O arrasto mínimo é {D(minimo)} e ocorre quando a velocidade é {minimo}.")
```
