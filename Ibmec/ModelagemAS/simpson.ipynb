Simpson(integração)
- Mais precisa: em vez de ligar os pontos com retas, o Simpson liga com parábolas. Parábola gruda melhor numa curva que uma reta, então o resultado é mais exato.

-Lembrar que so tem funcao simpson pra 1/3, não tem pra 3/8

-Simpson precisa de um numero PAR de intervalos, ou seja, IMPAR DE PONTOS, porque como faz com parabola, parabola usa 3 pontos.
- Se voce der numero PAR de pontos, ele faz com 3/8 



NOTAS DO CLAUE:
Simpson (integração)
- Mais precisa: em vez de ligar os pontos com retas, liga com PARÁBOLAS. 
  Parábola gruda melhor na curva que reta, então é mais exato.
- Simpson 1/3 precisa de número PAR de intervalos (= ÍMPAR de pontos), 
  porque cada parábola usa 3 pontos (fecha de 2 em 2 intervalos).
- No scipy só existe a função simpson (que é a 1/3). Não tem função pronta pra 3/8.
- Simpson 3/8: usa CÚBICAS (grupos de 4 pontos / 3 intervalos), pesos 1-3-3-1, 
  multiplica por 3h/8. Serve quando o nº de intervalos é múltiplo de 3, 
  ou pra "fechar" um intervalo que sobrou do 1/3.

Mesmo exemplo na mão (f(x)=x², 4 intervalos — par, OK):

I = (0,5/3)·[0 + 4(0,25) + 2(1) + 4(2,25) + 4] = 0,1667·[0 + 1 + 2 + 9 + 4] = 0,1667·16 = 2,667
Repara: deu exatamente o valor real (2,667), enquanto o trapézio deu 2,75. Simpson foi mais preciso.

```python
from scipy.integrate import simpson
from numpy import linspace
def f(x):
    return x**2

x = linspace(0, 2, 5)
I = simpson(f(x), x)

print(I)
```
