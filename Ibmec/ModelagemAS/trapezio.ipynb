Trapezio(integração) -> Integrar uma função é calcular a area embaixo da curva dela entre 2 limites (a e b). Fazemos isso aproximando por areas geometricas que conhecemos a area.

- A estratégia geral: divide o intervalo [a, b] em vários pedacinhos e aproxima a área de cada pedaço por uma forma conhecida. 

Exemplo na mão: integrar f(x) = x² de 0 a 2, com 4 intervalos.

```python
from scipy.integrate import trapezoid
from numpy import linspace

def f(x):
    return x**2

x = linspace(0, 2, 5)   # 5 pontos = 4 intervalos
I = trapezoid(f(x), x)
print(I)
```
