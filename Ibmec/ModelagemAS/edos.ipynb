EDOS (Equações diferenciais ordinarias) | EULER, HEUN E RUNGE-KUTTA | solve_ivp

-> É uma equação que te da a derivada de uma função e você quer descobrir a função real.

- A lógica de todos os métodos é a mesma: você está num ponto, calcula a inclinação ali (pela f), dá um passo de tamanho h na direção dessa inclinação, e chega num ponto novo e repete até chegar no final do intervalo x.

- Onde começar (a condição inicial, tipo x=0, y=1)
- Até onde ir (o x final, tipo x=4)
- O tamanho do passo (h, tipo 0,5)

A diferença entre Euler, Heun e Runge-Kutta é quão esperto é o cálculo da inclinação de cada passo:
- Euler: Calcula a inclinação só no ponto de partida e anda em linha reta com ela.
- Heun: O Heun melhora o Euler usando duas inclinações e tirando a média. Funciona em dois passos:
- Runge-Kutta: O mais preciso e mais usado. Usa quatro inclinações (k1, k2, k3, k4) em pontos estratégicos do passo e faz uma média ponderada delas.

A progressão é clara: quanto mais inclinações o método amostra, mais preciso ele é. Euler (1) é grosseiro, Heun (2) é bom, RK4 (4) é excelente. Esse é o tipo de comparação que a prova adora.



```python
from scipy.integrate import solve_ivp
from numpy import exp, linspace

def f(x, y):
    return 4*exp(0.8*x) - 0.5*y

sol = solve_ivp(f, [0, 4], [2], method='RK45', t_eval=linspace(0,4,10))
print(sol.y[0])
```
