Taylor é basicamente: 
- pega a função e deriva ela ate a ordem que pedir (geralmente pede ate a 3ª ordem) (f(x), df, df2,df3)
- substitui cada termo na formula do seu grau correspondente (ela da a formula de taylor na prova)
- No erro pra comparar depois faz a função incial com o valor de f(xfinal) - cada formula de taylor (variando de acordo com a ordem)
- Depois plota tudo pra gente ver em um grafico
- Na plotagem, plotamos um xx que tem o valor de linspace da função (xinicial, xfinal, 100 pontos), depois plotamos os pontos nas pontas da função e em seguida plotamos os pontos de taylor onde x dele = xfinal e y = valor de taylor de sua ordem.


```python
from numpy import *
from scipy.integrate import *
from scipy.interpolate import *
from matplotlib.pyplot import *

xinicial = 0
xfinal = 0.5
h = 0.5

f = lambda x: exp(x)
df = lambda x: exp(x)
df1 = lambda x: exp(x)
df2 = lambda x: exp(x)


taylor1 = f(xinicial) + df(xinicial)*h
taylor2 = taylor1 + (df1(xinicial)/2)*h**2
taylor3 = taylor2 + (df2(xinicial)/6)*h**3

erro1 = abs(f(xfinal) - taylor1)/f(xfinal)*100
erro2 = abs(f(xfinal)-  taylor2)/ f(xfinal)*100
erro3 = abs(f(xfinal) -  taylor3)/ f(xfinal)*100


print(f"O valor de f(xfinal) é de: ", f(xfinal))
print(f"O valor de taylor ordem 1 é de: ", taylor1)
print(f"O valor de taylor ordem 2 é de: ", taylor2)
print(f"O valor de taylor ordem 3 é de: ", taylor3)

print(f"O erro ordem 1 corresponde a {erro1}%")
print(f"O erro ordem 2 corresponde a {erro2}%")
print(f"O erro ordem 3 corresponde a {erro3}%")

xx = linspace(xinicial, xfinal, 100)
plot(xx, f(xx), 'g-', label = "funcao")
plot(xinicial, f(xinicial), 'r*', label = "ponto incial" )
plot(xfinal, f(xfinal), 'bs', label = "ponto final")
plot(xfinal, taylor1, 'yd', label = "taylor ordem 1")
plot(xfinal, taylor2, 'rd', label = "taylor ordem 2")
plot(xfinal, taylor3, 'kd', label = "taylor ordem 3")
legend()
grid()
```
