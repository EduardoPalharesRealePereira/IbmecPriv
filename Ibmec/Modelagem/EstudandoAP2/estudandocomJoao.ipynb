```python
from scipy.interpolate import *
from numpy import *
from matplotlib.pyplot import *

T= array([0,5,10,15,20,25,30])
gl = array([12.9,11.3,10.1,9.03,8.17,7.46,6.85])
resposta = polyfit(T,gl,6)
print(resposta)
resp2 = poly1d(resposta)
plot(T,gl, 'kd', label = "Dados")
plot(T, resp2(T), 'r', label = "Polinomio - ordem 6")
grid()
show()
Oitograus = resp2(8)
print("Em t = 8 graus = ",Oitograus)
```
