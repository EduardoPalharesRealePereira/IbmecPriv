```python
import scipy.interpolate as so

poly_pensar = so.lagrange([30, 120], [5.6, 22.4])
d_pensar = poly_pensar(110)

poly_frear = so.lagrange([30, 60, 120], [5, 21, 84.7])
d_frear = poly_frear(110)

print(d_frear)
print(d_pensar)
```

```python
import scipy.interpolate as si
import numpy as np
import matplotlib.pyplot as plt

x = np.array([3,4,5,6])
fx = np.array([0.3333, 0.25, 0.2, 0.1667])

P = si.lagrange(fx, x)

print(P(0.23))

plotar = P(0.23)

xx = np.linspace(0.3333, 0.1667, 100)

plt.plot(xx, P(xx), 'g', label="Polinomio")
plt.plot(fx, x, 'ro', label="Pontos")
plt.plot(0.23, plotar, 'y*', label="Resultado")
plt.legend()
plt.grid()
plt.show()


```

```python
import scipy.interpolate as si
import numpy as np
import matplotlib.pyplot as plt


hh = np.array([7.5, 8])
R = np.array([0.5940, 0.5653])

interpolacao = si.lagrange(hh, R)
print(interpolacao(7.67))
e1 = interpolacao(7.67)

hh2 = np.array([7, 7.5, 8])
R2 = np.array([0.6214, 0.5940, 0.5653])
interpolacao2 = si.lagrange(hh2, R2)
print(interpolacao2(7.67))
e2 = interpolacao2(7.67)


hh3 = np.array([7, 7.5, 8, 8.5])
R3 = np.array([0.6214, 0.5940, 0.5653, 0.5350])

interpolacao3 = si.lagrange(hh3, R3)
print(interpolacao3(7.67))
e3 = interpolacao3(7.67)

erro1 = e2 - e1
print("erro1 = ", erro1)
erro2 = e3 - e2
print("erro2 = ", erro2)

#falta a letra b da 3 aqui
```
