```python
import numpy as np
import scipy.integrate as si

g = 9.8
m = 68.1
c = 12.5

def f(t):
    v = ((g * m) / c) * (1 - np.exp(-(c / m) * t))
    return v

t = np.linspace(0, 30, 11)
I = si.trapezoid(f(t), t)

# Valor exato analítico: integral de v(t) de 0 a 30
exato = (g * m / c) * (30 + (m / c) * (np.exp(-(c / m) * 30) - 1))

erro = abs((exato - I) / exato)

print(f"Distância estimada (trapézio): {I:.4f} m")
print(f"Valor exato:                   {exato:.4f} m")
print(f"Erro relativo:                 {round(erro * 100, 2)} %")
```

# Página 24 — Regra 3/8 de Simpson + Regra 1/3 de Simpson (5 segmentos)

f(x) = 0,2 + 25x − 200x² + 675x³ − 900x⁴ + 400x⁵, de a=0 a b=0,8

Com 5 segmentos: aplica-se a regra 3/8 nos primeiros 3 segmentos e a regra 1/3 nos últimos 2.

```python
import numpy as np

def f(x):
    return 0.2 + 25*x - 200*x**2 + 675*x**3 - 900*x**4 + 400*x**5

a, b, n = 0, 0.8, 5
h = (b - a) / n
x = np.linspace(a, b, n + 1)  # 6 pontos: x0 a x5

# Regra 1/3 de Simpson: primeiros 2 segmentos (x0 a x2)
#daria pra usar simpson nessa aqui pq ela eh a de 1/3, claude n usou pra n confundir e fez as 2 na mão
I_13 = (h / 3) * (f(x[0]) + 4*f(x[1]) + f(x[2]))

# Regra 3/8 de Simpson: últimos 3 segmentos (x2 a x5)
I_38 = (3 * h / 8) * (f(x[2]) + 3*f(x[3]) + 3*f(x[4]) + f(x[5]))

I_total = I_13 + I_38
exato = 1.640533
erro = abs((exato - I_total) / exato) * 100

print(f"h = {h}")
print(f"Pontos x: {x}")
print(f"I (1/3 Simpson): {I_13:.6f}")
print(f"I (3/8 Simpson): {I_38:.6f}")
print(f"I total:         {I_total:.6f}")
print(f"Valor exato:     {exato}")
print(f"Erro relativo:   {erro:.4f} %")
```
